```python
import pandas as pdd

scores = pdd.Series([85, 92, 78, 90])
print("Scores:")
print(scores)
print(f"\nAverage: {scores.mean()}")
print(f"Maximum: {scores.max()}")
print(f"Minimum: {scores.min()}")
```
